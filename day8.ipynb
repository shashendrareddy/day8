```python
class Bike:
  wheels=2
  tyres =2
  colour =''
  height = 0
  width = 0
  mirrors =2
  engine = 0
  length = 0

  # parameterized constructor
  def __init__(self, clr: str, length : float, width: float, height: float):
    self.colour = clr
    self.length = length
    self.width = width
    self.height = height

  def manufacutre(self):
    print(f"""
            colour is {self.colour},
            height is {self.height} ,
            length is {self.length}
            """)


red_activa = Bike("red", 2.5, 1, 2)
red_activa.manufacutre()
```

```python

# simple/basic inheritance
class Parent:
  x=100
  f =5.5

  def __init__(self):
    print("this is parent class constructor")

class Child(Parent):
  x =200
  s ="hello"

  def __init__(self):
    super().__init__()
    print(super().x)
    print("this is child class constructor")

c= Child()
print(c.x, c.f)
```

```python

# multi level inheritance
class GrandParent:
  age =70
  children =2

  def __init__(self):
    print("this is grand parent class constructor")

class Parent(GrandParent):
  age =45
  def __init__(self):
    print(super().age)
    print("this is parent class constructor")

class Child(Parent):
  age =16
  def __init__(self):
    print(super().age)
    super().__init__()
    print("this is child class constructor")

c = Child()
print(c.age)
print(c.children)
```

```python

# multiple inheritance
class Parent1:
  money =1000
  age = 45
  def __init__(self):
    print("this is parent1 class constructor")

class Parent2:
  money = 2000
  age = 43
  sum =100
  def __init__(self):
    print("this is parent2 class constructor")

class Child(Parent2, Parent1):
  money = 0
  def __init__(self):
    super().__init__()
    print("this is child class constructor")


c = Child()
print(c.money)
print(c.age)
print(c.sum)
```

```python
# heirarichal inheritance
class Parent:
  age =50
  money = 5000
  def __init__(self):
    print("this is parent class constructor")

class Child1(Parent):
  age = 20
  money =100
  def __init__(self):
    self.money = 1000
    print(super().age)
    print("this is child1 class constructor")

class Child2(Parent):
  age = 18
  money =80
  def __init__(self):
    print(super().age)
    print("this is child2 class constructor")

c1= Child1()
print(c1.money)
c2= Child2()
print(c2.money)
```
